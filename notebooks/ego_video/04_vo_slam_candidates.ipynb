# 04 VO And SLAM Candidates

Planning notebook for heavier camera-trajectory methods after the classical baseline is understood.

## Candidate Order

1. ORB-SLAM3
2. OpenVSLAM
3. DROID-SLAM
4. DPVO

Why:
- ORB-SLAM3 is the best fisheye-first classical choice.
- OpenVSLAM is a good engineering comparison.
- DROID-SLAM is stronger but heavier.
- DPVO is interesting as a learned VO baseline.
